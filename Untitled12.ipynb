```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import blob_log
from skimage.color import rgb2gray
from scipy.optimize import minimize
```

```python
# 1. Load the sunflower field image
image_path = '/content/the_berry_farms_sunflower_field.jpeg'
im = cv.imread(image_path, cv.IMREAD_REDUCED_COLOR_4)

if im is None:
    print(f"Error: Could not load image from {image_path}. Check the file path.")
else:
    # Convert image to grayscale and normalize values to [0, 1]
    image_gray = rgb2gray(cv.cvtColor(im, cv.COLOR_BGR2RGB))

    # 2. Define Scale-Space Parameters for LoG Blob Detection
    min_sigma = 1
    max_sigma = 30
    num_sigma = 10
    threshold = 0.1  # Adjust sensitivity depending on image contrast

    print(f"--- Scale-Space Configuration ---")
    print(f"Minimum Sigma (min_sigma): {min_sigma}")
    print(f"Maximum Sigma (max_sigma): {max_sigma}")
    print(f"Number of Sigma steps (num_sigma): {num_sigma}")

    # Perform Laplacian of Gaussian (LoG) blob detection
    # Returns an array of coordinates and scales: [y, x, sigma]
    blobs_log = blob_log(
        image_gray,
        min_sigma=min_sigma,
        max_sigma=max_sigma,
        num_sigma=num_sigma,
        threshold=threshold
    )

    # Convert sigma to circle radius: r = sigma * sqrt(2)
    blobs_log[:, 2] = blobs_log[:, 2] * np.sqrt(2)

    # 3. Report parameters of the largest circles
    if len(blobs_log) > 0:
        # Sort blobs by radius in descending order
        sorted_blobs = blobs_log[np.argsort(blobs_log[:, 2])[::-1]]

        print("\n--- Top 5 Largest Detected Circles ---")
        for i in range(min(5, len(sorted_blobs))):
            y, x, r = sorted_blobs[i]
            print(f"Circle {i+1}: Center X = {x:.2f}, Center Y = {y:.2f}, Radius (r) = {r:.2f}")

        # 4. Visualize and draw circles on the image
        fig, ax = plt.subplots(figsize=(12, 8))
        ax.imshow(cv.cvtColor(im, cv.COLOR_BGR2RGB))

        for blob in blobs_log:
            y, x, r = blob
            c = plt.Circle((x, y), r, color='yellow', linewidth=1.2, fill=False)
            ax.add_patch(c)

        plt.title("Sunflower Field Blob Detection (Laplacian of Gaussians)")
        plt.axis('off')
        plt.tight_layout()
        plt.show()
    else:
        print("No blobs detected. Try lowering the threshold value.")
```

```python
"""
EN3160 Assignment 2 - Q2
RANSAC (coded from scratch) to fit a line, then a circle to the remnant.
Also shows what happens if the circle is fitted first (part d).

Line  : unit normal [a, b], offset d  ->  a*x + b*y = d,  a^2 + b^2 = 1
Circle: centre (cx, cy), radius r
"""

# ----------------------------------------------------------------------------
# Noisy point set (Listing 1)
# ----------------------------------------------------------------------------
np.random.seed(0)
N = 100
half_n = N // 2
r = 10
x0_gt, y0_gt = 2, 3                       # circle centre
s = r / 16
t = np.random.uniform(0, 2 * np.pi, half_n)
n = s * np.random.randn(half_n)
x = x0_gt + (r + n) * np.cos(t)
y = y0_gt + (r + n) * np.sin(t)
X_circ = np.hstack((x.reshape(half_n, 1), y.reshape(half_n, 1)))

s = 1.
m, b = -1, 2
x = np.linspace(-12, 12, half_n)
y = m * x + b + s * np.random.randn(half_n)
X_line = np.hstack((x.reshape(half_n, 1), y.reshape(half_n, 1)))

X = np.vstack((X_circ, X_line))           # all points

# ----------------------------------------------------------------------------
# RANSAC settings
# ----------------------------------------------------------------------------
# Line noise sigma = 1.0 (normal direction) -> threshold 1.5 keeps ~87% of true inliers.
LINE_THRESH = 1.5
LINE_MIN_CONSENSUS = 30          # true line has 50 pts; need a clear majority of them
# Circle radial noise sigma = r/16 = 0.625 -> threshold ~2 sigma.
CIRCLE_THRESH = 1.25
CIRCLE_MIN_CONSENSUS = 25        # true circle has 50 pts; some are lost to the line step

# Iterations from  N = log(1-p) / log(1 - w^s)
P_SUCCESS, W_GUESS = 0.999, 0.35


def n_iters(sample_size, w=W_GUESS, p=P_SUCCESS):
    return int(np.ceil(np.log(1 - p) / np.log(1 - w ** sample_size)))


# ----------------------------------------------------------------------------
# Line model
# ----------------------------------------------------------------------------
def line_from_2pts(p1, p2):
    """Unit normal (a, b) and offset d so that a*x + b*y = d."""
    dx, dy = p2 - p1
    norm = np.hypot(dx, dy)
    if norm < 1e-9:
        return None
    a, b = -dy / norm, dx / norm          # ||[a, b]|| = 1 by construction
    return a, b, a * p1[0] + b * p1[1]


def line_error(P, model):
    a, b, d = model
    return np.abs(P @ np.array([a, b]) - d)   # normal distance to the line


def fit_line_tls(P):
    """Total least squares line fit with the constraint ||[a,b]|| = 1 (via SVD)."""
    c = P.mean(axis=0)
    _, _, Vt = np.linalg.svd(P - c)
    a, b = Vt[-1]                          # direction of smallest variance = normal
    return a, b, a * c[0] + b * c[1]


def ransac_line(P, thresh, min_consensus, iters, rng):
    best = dict(count=-1)
    for _ in range(iters):
        idx = rng.choice(len(P), 2, replace=False)
        model = line_from_2pts(P[idx[0]], P[idx[1]])
        if model is None:
            continue
        inl = line_error(P, model) < thresh
        cnt = inl.sum()
        if cnt > best['count']:
            best = dict(count=cnt, model=model, inliers=inl, sample=P[idx])
    if best['count'] < min_consensus:
        raise RuntimeError('Line RANSAC: consensus too small')
    best['refined'] = fit_line_tls(P[best['inliers']])
    return best


# ----------------------------------------------------------------------------
# Circle model
# ----------------------------------------------------------------------------
def circle_from_3pts(p1, p2, p3):
    A = 2 * np.array([[p2[0] - p1[0], p2[1] - p1[1]],
                      [p3[0] - p1[0], p3[1] - p1[1]]])
    rhs = np.array([p2 @ p2 - p1 @ p1, p3 @ p3 - p1 @ p1])
    if abs(np.linalg.det(A)) < 1e-9:       # collinear points
        return None
    c = np.linalg.solve(A, rhs)
    return c[0], c[1], np.linalg.norm(p1 - c)


def circle_error(P, model):
    cx, cy, rad = model
    return np.abs(np.hypot(P[:, 0] - cx, P[:, 1] - cy) - rad)   # radial error


def fit_circle_opt(P, init):
    """Least-squares circle fit over the inliers using scipy.optimize.minimize."""
    def cost(p):
        return np.sum((np.hypot(P[:, 0] - p[0], P[:, 1] - p[1]) - p[2]) ** 2)
    res = minimize(cost, x0=np.array(init), method='BFGS')
    return tuple(res.x)


def ransac_circle(P, thresh, min_consensus, iters, rng):
    best = dict(count=-1)
    for _ in range(iters):
        idx = rng.choice(len(P), 3, replace=False)
        model = circle_from_3pts(*P[idx])
        if model is None:
            continue
        inl = circle_error(P, model) < thresh
        cnt = inl.sum()
        if cnt > best['count']:
            best = dict(count=cnt, model=model, inliers=inl, sample=P[idx])
    if best['count'] < min_consensus:
        raise RuntimeError('Circle RANSAC: consensus too small')
    best['refined'] = fit_circle_opt(P[best['inliers']], best['model'])
    return best


# ----------------------------------------------------------------------------
# (a) line with RANSAC,  (b) remove consensus, circle with RANSAC on the remnant
# ----------------------------------------------------------------------------
rng = np.random.default_rng(0)
line_res = ransac_line(X, LINE_THRESH, LINE_MIN_CONSENSUS, n_iters(2), rng)
remnant = X[~line_res['inliers']]
circ_res = ransac_circle(remnant, CIRCLE_THRESH, CIRCLE_MIN_CONSENSUS, n_iters(3), rng)

a, b_, d = line_res['refined']
cx, cy, rad = circ_res['refined']
print('--- Line (RANSAC + TLS refinement) ---')
print(f'  a={a:.4f}, b={b_:.4f}, d={d:.4f}   |[a,b]|={np.hypot(a, b_):.4f}')
print(f'  inliers: {line_res["count"]}')
gt_n = np.array([1, 1]) / np.sqrt(2)       # ground truth y = -x + 2  ->  x + y = 2
print(f'  ground truth (normalised): a=+-{gt_n[0]:.4f}, b=+-{gt_n[1]:.4f}, d=+-{2 / np.sqrt(2):.4f}')
print('--- Circle (RANSAC + optimiser refinement) ---')
print(f'  centre=({cx:.3f}, {cy:.3f}), radius={rad:.3f}   inliers: {circ_res["count"]}')
print(f'  ground truth: centre=({x0_gt}, {y0_gt}), radius={r}')


# ----------------------------------------------------------------------------
# (c) plot
# ----------------------------------------------------------------------------
def line_pts(model, half_len=40):
    a, b, d = model
    p0 = d * np.array([a, b])               # closest point to origin
    t_ = np.array([-b, a])                  # direction along the line
    return np.vstack((p0 - half_len * t_, p0 + half_len * t_))


fig, ax = plt.subplots(1, 1, figsize=(9, 9))
ax.scatter(X[:, 0], X[:, 1], s=18, label='All points', zorder=1)
li = X[line_res['inliers']]
ax.scatter(li[:, 0], li[:, 1], s=18, c='y', label='Line inliers', zorder=2)
ls = line_res['sample']
ax.scatter(ls[:, 0], ls[:, 1], marker='x', c='r', s=50, label='Best sample for line', zorder=4)
ci = remnant[circ_res['inliers']]
ax.scatter(ci[:, 0], ci[:, 1], s=18, c='c', label='Circle inliers', zorder=2)
cs = circ_res['sample']
ax.scatter(cs[:, 0], cs[:, 1], c='r', s=30, label='Best sample for circle', zorder=4)

ax.add_patch(plt.Circle((x0_gt, y0_gt), r, color='g', fill=False, label='Ground truth circle'))
ax.add_patch(plt.Circle((cx, cy), rad, color='m', fill=False, label='RANSAC circle (refined)'))
sx, sy, sr = circ_res['model']
ax.add_patch(plt.Circle((sx, sy), sr, color='k', fill=False, ls='--', lw=1,
                        label='Circle from best sample'))

xs = np.array([-14, 14])
ax.plot(xs, m * xs + b, color='g', lw=1, label='Ground truth line')
gl = line_pts(line_res['refined'])
ax.plot(gl[:, 0], gl[:, 1], color='m', lw=1, label='RANSAC line (refined)')
sl = line_pts(line_res['model'])
ax.plot(sl[:, 0], sl[:, 1], color='k', ls='--', lw=1, label='Line from best sample')

ax.set_xlim(-14, 14); ax.set_ylim(-13, 16); ax.set_aspect('equal')
ax.legend(fontsize=8, loc='upper right')
ax.set_title('Line then circle fitting with RANSAC')
plt.savefig('q2_line_then_circle.png', dpi=150, bbox_inches='tight')

# ----------------------------------------------------------------------------
# (d) what if the circle is fitted FIRST?
# ----------------------------------------------------------------------------
# The line points are 50 (almost collinear) points. A circle of large radius
# passes within the threshold of a long stretch of them, so a circle through
# line points (plus a few circle points) can collect as many inliers as the true
# circle, and the best model found is often wrong. Run it and compare.
rng2 = np.random.default_rng(1)
try:
    wrong = ransac_circle(X, CIRCLE_THRESH, CIRCLE_MIN_CONSENSUS, n_iters(3), rng2)
    wx, wy, wr = wrong['refined']
    print('--- Circle fitted first ---')
    print(f'  centre=({wx:.2f}, {wy:.2f}), radius={wr:.2f}, inliers={wrong["count"]}  '
          f'(truth: ({x0_gt}, {y0_gt}), {r})')
    fig2, ax2 = plt.subplots(1, 1, figsize=(8, 8))
    ax2.scatter(X[:, 0], X[:, 1], s=18, label='All points')
    wi = X[wrong['inliers']]
    ax2.scatter(wi[:, 0], wi[:, 1], s=18, c='c', label='"Circle" inliers')
    ax2.add_patch(plt.Circle((x0_gt, y0_gt), r, color='g', fill=False, label='Ground truth circle'))
    ax2.add_patch(plt.Circle((wx, wy), wr, color='m', fill=False, label='Circle fitted first'))
    ax2.set_xlim(-14, 14); ax2.set_ylim(-13, 16); ax2.set_aspect('equal')
    ax2.legend(fontsize=8); ax2.set_title('Fitting the circle first')
    plt.savefig('q2_circle_first.png', dpi=150, bbox_inches='tight')
except RuntimeError as e:
    print('Circle first:', e)

plt.show()
```
